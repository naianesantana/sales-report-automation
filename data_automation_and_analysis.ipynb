# Automação de Sistemas e Processos com Python

### Desafio:

Todos os dias, o nosso sistema atualiza as vendas do dia anterior.
O seu trabalho diário, como analista, é enviar um e-mail para a diretoria, assim que começar a trabalhar, com o faturamento e a quantidade de produtos vendidos no dia anterior

E-mail da diretoria: seugmail+diretoria@gmail.com<br>
Local onde o sistema disponibiliza as vendas do dia anterior: https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga

Para resolver isso, vamos usar o pyautogui, uma biblioteca de automação de comandos do mouse e do teclado


Referência do pyautogui: https://pyautogui.readthedocs.io/en/latest/quickstart.html

```python
#Bibliotecas que irei utilizar: pyautogui (possibilita o controle do mouse, teclado e tela do dispositivo) 
# | pandas (manipulação dos dados) 
# | openpyxl (fornece um plus para o pandas, possibilitando o uso de base de dados do Excel)

# Comando para instalação das bibliotecas via terminal > cmd:
# pip install pyautogui pandas openpyxl

# pyautogui.click > clica com o mouse
# pyautogui.write > escreve um texto
# pyautogui.press > pressiona 1 tecla 
# pyautogui.hotkey > combinação de teclas (atalhos)
```

```python
import pyautogui
import time

pyautogui.PAUSE = 1.5
link = "https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga"

# 1º Passo: Entrar no sistema da empresa (link do Drive)
### Sub-passos: Acessar o menu Windows
pyautogui.press('win')

### > Pesquisar pelo navegador
pyautogui.write('brave')
pyautogui.press('enter')

### > Acessar o navegador
pyautogui.click(x=764, y=373, clicks=2)
pyautogui.hotkey('ctrl', 't')


### > Colar o link do Drive na barra de pesquisa e realizar a busca
pyautogui.write(link)
time.sleep(2)
pyautogui.press('enter')
```

```python
# 2º Passo: Navegar no sistema para encontrar a base de dados
time.sleep(10)
pyautogui.click(x=412, y=331, clicks=2)

# 3º Passo: Exportar a base de dados (baixar o arquivo)
time.sleep(5)
pyautogui.click(x=506, y=323)
pyautogui.click(x=569, y=395)
time.sleep(5)
```

### Vamos agora ler o arquivo baixado para pegar os indicadores

- Faturamento
- Quantidade de Produtos

```python
import pandas
# 4º Passo: Calcular os indicadores (faturamento e quantidade de produtos vendidos)
### Sub-passos: Abrir a base de dados
caminho = R"C:\Caminho\Para\Localizar\Arquivo\baixado.xlsx"
db = pandas.read_excel(caminho)

### Visualizar as informações existentes
display(db)

### Somar o faturamento
faturamento = db["Valor Final"].sum()

### Somar a quantidade de produtos vendidos
qtd_vendida = db["Quantidade"].sum()

#print(faturamento)
#print(qtd_vendida)
```

### Vamos agora enviar um e-mail pelo gmail

```python
#Importei a biblioteca pyperclip pois estava enfrentando dificuldades na digitação de letras acentuadas pelo pyautogui.write(ele não digitava elas)
import pyperclip

# 5º Passo: Enviar as informações por e-mail
## Acessando o Gmail 
email = "https://mail.google.com"
endereco_email = "seuemail+codigopython@gmail.com"
pyautogui.hotkey('ctrl','t')
pyautogui.write(email)
pyautogui.press('enter')
time.sleep(4)

## Escrevendo uma nova mensagem de e-mail
pyautogui.click(x=36, y=180)
pyautogui.write(endereco_email)
time.sleep(3)
pyautogui.press('tab', presses=2)
pyperclip.copy('Relatório de Vendas Atualizado')
pyautogui.hotkey('ctrl','v')
time.sleep(3)
pyautogui.press('tab')

## Formatando os valores dos indicadores
fat_inter = f'{faturamento:,.2f}'
fat_final = fat_inter.replace(',','_').replace('.',',').replace('_','.')

qtd_vendida_final = f'{qtd_vendida:,}'.replace(',','.')

## Escrevendo a mensagem de e-mail
corpo_email = f'''Olá, João,

Espero que este e-mail o encontre bem.

Segue os indicadores de vendas atualizado referente ao período vigente.

Destaques do Período:

- Faturamento Total: R$ {fat_final}

- Volume de Vendas: {qtd_vendida_final} produtos vendidos

Para mais detalhes sobre os indicadores de desempenho e o detalhamento por região/vendedor, por favor, consulte o documento anexado.

Atenciosamente,

Naiane Santana
'''
pyperclip.copy(corpo_email)
pyautogui.hotkey('ctrl','v')

## Enviando a mensagem
pyautogui.hotkey('ctrl','enter')
```
